```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. LOAD DATASET & EXAMINE FEATURES
# Load your cleaned dataset
df = pd.read_csv("cleaned_dataset.csv")

# Display basic information about the dataset
print("--- Dataset Information ---")

print("--- Dataset Shape ---")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

print("\n--- Summary Statistics (Numerical Features) ---")
print(df.describe())

print("\n--- Summary Statistics (Categorical Features) ---")
print(df.describe(include=['O']))


# 2. IDENTIFY TRENDS, DISTRIBUTIONS & RELATIONSHIPS
print("\n--- Skewness of Numerical Features ---")
print(df.skew(numeric_only=True))

print("\n--- Correlation Matrix ---")
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)


# 3. OUTLIER DETECTION USING IQR (INTERQUARTILE RANGE)
def detect_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    print(f"Feature '{column}': {len(outliers)} outliers detected (Boundaries: [{lower_bound:.2f}, {upper_bound:.2f}])")
    return outliers

# Example: Check for outliers in a numeric column (replace 'Sales' with your column)
# detect_outliers_iqr(df, 'Sales')


# 4. ANSWER KEY BUSINESS QUESTIONS (EXAMPLES)
# Example Question 1: What is the average value by category?
# print(df.groupby('Category')['Sales'].mean())

# Example Question 2: What are the top 5 performing segments?
# print(df['Segment'].value_counts().head(5))
```
